```python
import json
from pathlib import Path

import pandas as pd
```

```python
files = list(Path("../data").glob("*.json"))

print("Antal JSON-filer:", len(files))
```

```python
# Läser in flygdata från JSON-filer och skapar en DataFrame (df)

rows = []

for file in files:
    with file.open(encoding="utf-8") as f:
        data = json.load(f)

    direction, airport, flight_date = file.stem.split("_", 2)
    flights = data.get("flights", []) if isinstance(data, dict) else data

    for flight in flights:
        if isinstance(flight, dict):
            row = flight.copy()
            row["airport"] = airport
            row["flight_date"] = flight_date
            row["direction"] = direction
            rows.append(row)

df = pd.DataFrame(rows)
```
